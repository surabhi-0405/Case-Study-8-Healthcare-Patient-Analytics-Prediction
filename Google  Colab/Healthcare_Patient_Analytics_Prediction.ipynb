```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries imported successfully.")
```

```python
import os

print("Files available in Colab:")
for file in os.listdir('/content'):
    print(file)
```

```python
# Load healthcare datasets

patients = pd.read_csv('/content/patients.csv')
admissions = pd.read_csv('/content/admissions.csv')
doctors = pd.read_csv('/content/doctors.csv')
treatments = pd.read_csv('/content/treatments.csv')
vitals = pd.read_csv('/content/vitals.csv')

print("Datasets loaded successfully.\n")

print("Patients   :", patients.shape)
print("Admissions :", admissions.shape)
print("Doctors    :", doctors.shape)
print("Treatments :", treatments.shape)
print("Vitals     :", vitals.shape)
```

```python
# Display structure of all datasets

datasets = {
    "Patients": patients,
    "Admissions": admissions,
    "Doctors": doctors,
    "Treatments": treatments,
    "Vitals": vitals
}

for name, df in datasets.items():
    print("=" * 70)
    print(f"{name} Dataset")
    print("=" * 70)
    print("Columns:")
    print(df.columns.tolist())
    print("\nData Types:")
    print(df.dtypes)
    print()
```

```python
# Check missing values in all datasets

for name, df in datasets.items():
    print("=" * 70)
    print(f"{name} - Missing Values")
    print("=" * 70)

    missing = df.isnull().sum()
    missing_percent = (df.isnull().sum() / len(df) * 100).round(2)

    missing_summary = pd.DataFrame({
        'Missing_Count': missing,
        'Missing_Percentage': missing_percent
    })

    print(missing_summary)
    print()
```

```python
# Check duplicate records and primary key duplicates

primary_keys = {
    "Patients": "patient_id",
    "Admissions": "admission_id",
    "Doctors": "doctor_id",
    "Treatments": "treatment_id",
    "Vitals": "vital_id"
}

for name, df in datasets.items():
    pk = primary_keys[name]

    print("=" * 70)
    print(f"{name}")
    print("=" * 70)
    print(f"Full duplicate rows      : {df.duplicated().sum()}")
    print(f"Duplicate {pk} values    : {df[pk].duplicated().sum()}")
    print(f"Unique {pk} values       : {df[pk].nunique()}")
    print(f"Total rows               : {len(df)}")
    print()
```

```python
# Foreign Key Validation

print("=" * 70)
print("FOREIGN KEY VALIDATION")
print("=" * 70)

# Admissions → Patients
invalid_patient_fk = ~admissions['patient_id'].isin(patients['patient_id'])
print(f"\nAdmissions → Patients")
print(f"Invalid patient_id references: {invalid_patient_fk.sum()}")

# Admissions → Doctors
invalid_doctor_fk = ~admissions['doctor_id'].isin(doctors['doctor_id'])
print(f"\nAdmissions → Doctors")
print(f"Invalid doctor_id references: {invalid_doctor_fk.sum()}")

# Vitals → Admissions
invalid_vitals_fk = ~vitals['admission_id'].isin(admissions['admission_id'])
print(f"\nVitals → Admissions")
print(f"Invalid admission_id references: {invalid_vitals_fk.sum()}")

# Treatments → Admissions
invalid_treatments_fk = ~treatments['admission_id'].isin(admissions['admission_id'])
print(f"\nTreatments → Admissions")
print(f"Invalid admission_id references: {invalid_treatments_fk.sum()}")

print("\nForeign key validation completed.")
```

```python
# Convert date and datetime columns

patients['dob'] = pd.to_datetime(patients['dob'])

admissions['admission_date'] = pd.to_datetime(admissions['admission_date'])
admissions['discharge_date'] = pd.to_datetime(admissions['discharge_date'])

treatments['treatment_date'] = pd.to_datetime(treatments['treatment_date'])

vitals['recorded_time'] = pd.to_datetime(vitals['recorded_time'])

print("Date and datetime columns converted successfully.\n")

print("Patients dob:", patients['dob'].dtype)
print("Admissions admission_date:", admissions['admission_date'].dtype)
print("Admissions discharge_date:", admissions['discharge_date'].dtype)
print("Treatments treatment_date:", treatments['treatment_date'].dtype)
print("Vitals recorded_time:", vitals['recorded_time'].dtype)
```

```python
# Validate admission and discharge dates

admissions['length_of_stay'] = (
    admissions['discharge_date'] - admissions['admission_date']
).dt.days

print("=" * 70)
print("ADMISSION DATE VALIDATION")
print("=" * 70)

print("\nInvalid admission periods (discharge before admission):")
print((admissions['length_of_stay'] < 0).sum())

print("\nLength of Stay Summary:")
print(admissions['length_of_stay'].describe())

print("\nMinimum Length of Stay:", admissions['length_of_stay'].min(), "days")
print("Maximum Length of Stay:", admissions['length_of_stay'].max(), "days")
print("Average Length of Stay:", round(admissions['length_of_stay'].mean(), 2), "days")
```

```python
# Validate treatment dates against admission and discharge dates

treatment_check = treatments.merge(
    admissions[['admission_id', 'admission_date', 'discharge_date']],
    on='admission_id',
    how='left'
)

before_admission = (
    treatment_check['treatment_date'] < treatment_check['admission_date']
).sum()

after_discharge = (
    treatment_check['treatment_date'] > treatment_check['discharge_date']
).sum()

print("=" * 70)
print("TREATMENT DATE VALIDATION")
print("=" * 70)

print("\nTreatments before admission:", before_admission)
print("Treatments after discharge :", after_discharge)

print("\nTreatment date range:")
print("Minimum:", treatments['treatment_date'].min())
print("Maximum:", treatments['treatment_date'].max())
```

```python
# Validate vital-record dates against admission and discharge dates

vitals_check = vitals.merge(
    admissions[['admission_id', 'admission_date', 'discharge_date']],
    on='admission_id',
    how='left'
)

vitals_before_admission = (
    vitals_check['recorded_time'].dt.date <
    vitals_check['admission_date'].dt.date
).sum()

vitals_after_discharge = (
    vitals_check['recorded_time'].dt.date >
    vitals_check['discharge_date'].dt.date
).sum()

print("=" * 70)
print("VITALS DATE VALIDATION")
print("=" * 70)

print("\nVitals before admission:", vitals_before_admission)
print("Vitals after discharge :", vitals_after_discharge)

print("\nVitals recording date range:")
print("Minimum:", vitals['recorded_time'].min())
print("Maximum:", vitals['recorded_time'].max())
```

```python
# Validate vital sign ranges

print("=" * 70)
print("VITAL SIGN VALIDATION")
print("=" * 70)

print("\nHeart Rate:")
print("Minimum:", vitals['heart_rate'].min())
print("Maximum:", vitals['heart_rate'].max())

print("\nOxygen Level:")
print("Minimum:", vitals['oxygen_level'].min())
print("Maximum:", vitals['oxygen_level'].max())

print("\nTemperature:")
print("Minimum:", vitals['temperature'].min())
print("Maximum:", vitals['temperature'].max())

print("\nBlood Pressure - Sample Values:")
print(vitals['blood_pressure'].value_counts().head(10))
```

```python
# Split blood pressure into systolic and diastolic values

vitals[['systolic_bp', 'diastolic_bp']] = (
    vitals['blood_pressure']
    .str.split('/', expand=True)
    .astype(int)
)

print("=" * 70)
print("BLOOD PRESSURE VALIDATION")
print("=" * 70)

print("\nSystolic BP:")
print("Minimum:", vitals['systolic_bp'].min())
print("Maximum:", vitals['systolic_bp'].max())

print("\nDiastolic BP:")
print("Minimum:", vitals['diastolic_bp'].min())
print("Maximum:", vitals['diastolic_bp'].max())

print("\nMissing parsed values:")
print(vitals[['systolic_bp', 'diastolic_bp']].isnull().sum())
```

```python
# Check categorical values

print("=" * 70)
print("CATEGORICAL VALUE VALIDATION")
print("=" * 70)

print("\n--- Patient Gender ---")
print(patients['gender'].value_counts(dropna=False))

print("\n--- Chronic Conditions ---")
print(patients['chronic_conditions'].value_counts(dropna=False).head(15))

print("\n--- Admission Diagnoses ---")
print(admissions['diagnosis'].value_counts(dropna=False))

print("\n--- Doctor Specializations ---")
print(doctors['specialization'].value_counts(dropna=False))

print("\n--- Treatment Procedures ---")
print(treatments['procedure'].value_counts(dropna=False).head(15))

print("\n--- Treatment Medications ---")
print(treatments['medication'].value_counts(dropna=False).head(15))
```

```python
# Handle missing chronic condition values

patients['chronic_conditions'] = patients['chronic_conditions'].fillna(
    'No Recorded Condition'
)

print("=" * 70)
print("MISSING VALUE HANDLING")
print("=" * 70)

print("\nMissing values after cleaning:")
print(patients['chronic_conditions'].isnull().sum())

print("\nUpdated Chronic Conditions:")
print(patients['chronic_conditions'].value_counts())
```

```python
# Calculate patient age

reference_date = admissions['admission_date'].max()

patients['age'] = (
    (reference_date - patients['dob']).dt.days / 365.25
).astype(int)

print("=" * 70)
print("PATIENT AGE VALIDATION")
print("=" * 70)

print("\nMinimum Age:", patients['age'].min())
print("Maximum Age:", patients['age'].max())
print("Average Age:", round(patients['age'].mean(), 2))

print("\nAge Summary:")
print(patients['age'].describe())
```

```python
# Create admission-level features

admissions = admissions.sort_values(
    ['patient_id', 'admission_date']
).copy()

admissions['previous_admissions'] = (
    admissions.groupby('patient_id').cumcount()
)

print("=" * 70)
print("ADMISSION-LEVEL FEATURES")
print("=" * 70)

print("\nPrevious Admissions Summary:")
print(admissions['previous_admissions'].describe())

print("\nSample:")
print(
    admissions[
        ['admission_id', 'patient_id', 'admission_date',
         'length_of_stay', 'previous_admissions']
    ].head(10)
)
```

```python
# Create chronic condition indicator

patients['has_chronic_condition'] = (
    patients['chronic_conditions'] != 'No Recorded Condition'
).astype(int)

print("=" * 70)
print("CHRONIC CONDITION FEATURE")
print("=" * 70)

print("\nPatients with chronic conditions:")
print(patients['has_chronic_condition'].sum())

print("\nPatients without recorded chronic conditions:")
print((patients['has_chronic_condition'] == 0).sum())

print("\nFeature distribution:")
print(patients['has_chronic_condition'].value_counts())
```

```python
# Aggregate vital signs at admission level

vitals_agg = (
    vitals.groupby('admission_id')
    .agg(
        avg_heart_rate=('heart_rate', 'mean'),
        min_heart_rate=('heart_rate', 'min'),
        max_heart_rate=('heart_rate', 'max'),
        avg_oxygen_level=('oxygen_level', 'mean'),
        min_oxygen_level=('oxygen_level', 'min'),
        avg_temperature=('temperature', 'mean'),
        max_temperature=('temperature', 'max'),
        avg_systolic_bp=('systolic_bp', 'mean'),
        avg_diastolic_bp=('diastolic_bp', 'mean'),
        vital_readings=('vital_id', 'count')
    )
    .reset_index()
)

print("=" * 70)
print("VITALS AGGREGATION")
print("=" * 70)

print("\nAggregated shape:", vitals_agg.shape)

print("\nSample:")
print(vitals_agg.head())

print("\nMissing values:")
print(vitals_agg.isnull().sum())
```

```python
# Aggregate treatment information at admission level

treatments_agg = (
    treatments.groupby('admission_id')
    .agg(
        treatment_count=('treatment_id', 'count'),
        procedure_count=('procedure', 'nunique'),
        medication_count=('medication', 'nunique')
    )
    .reset_index()
)

print("=" * 70)
print("TREATMENT AGGREGATION")
print("=" * 70)

print("\nAggregated shape:", treatments_agg.shape)

print("\nSample:")
print(treatments_agg.head())

print("\nMissing values:")
print(treatments_agg.isnull().sum())
```

```python
# Create the admission-level analytics dataset

analytics_df = (
    admissions
    .merge(
        patients[
            ['patient_id', 'age', 'gender',
             'chronic_conditions', 'has_chronic_condition']
        ],
        on='patient_id',
        how='left'
    )
    .merge(
        vitals_agg,
        on='admission_id',
        how='left'
    )
    .merge(
        treatments_agg,
        on='admission_id',
        how='left'
    )
)

print("=" * 70)
print("ADMISSION-LEVEL ANALYTICS DATASET")
print("=" * 70)

print("\nShape:", analytics_df.shape)

print("\nMissing values:")
print(analytics_df.isnull().sum())

print("\nSample:")
print(analytics_df.head())
```

```python
# Numerical feature summary

numeric_columns = [
    'age',
    'length_of_stay',
    'previous_admissions',
    'avg_heart_rate',
    'avg_oxygen_level',
    'avg_temperature',
    'avg_systolic_bp',
    'avg_diastolic_bp',
    'vital_readings',
    'treatment_count'
]

print("=" * 70)
print("NUMERICAL FEATURE SUMMARY")
print("=" * 70)

display(
    analytics_df[numeric_columns]
    .describe()
    .round(2)
)
```

```python
# Admission distribution by diagnosis

diagnosis_counts = (
    analytics_df['diagnosis']
    .value_counts()
    .reset_index()
)

diagnosis_counts.columns = ['diagnosis', 'admission_count']

print("=" * 70)
print("ADMISSIONS BY DIAGNOSIS")
print("=" * 70)

display(diagnosis_counts)

print("\nTotal Admissions:", diagnosis_counts['admission_count'].sum())
```

```python
# Monthly admission trend

analytics_df['admission_month'] = (
    analytics_df['admission_date']
    .dt.to_period('M')
    .astype(str)
)

monthly_admissions = (
    analytics_df['admission_month']
    .value_counts()
    .sort_index()
    .reset_index()
)

monthly_admissions.columns = ['admission_month', 'admission_count']

print("=" * 70)
print("MONTHLY ADMISSION TREND")
print("=" * 70)

display(monthly_admissions)

print("\nTotal Admissions:", monthly_admissions['admission_count'].sum())
```

```python
# Admission distribution by gender

gender_admissions = (
    analytics_df['gender']
    .value_counts()
    .reset_index()
)

gender_admissions.columns = ['gender', 'admission_count']

print("=" * 70)
print("ADMISSIONS BY GENDER")
print("=" * 70)

display(gender_admissions)

print("\nTotal Admissions:", gender_admissions['admission_count'].sum())
```

```python
# Admissions by chronic condition

chronic_admissions = (
    analytics_df['chronic_conditions']
    .value_counts()
    .reset_index()
)

chronic_admissions.columns = ['chronic_condition', 'admission_count']

print("=" * 70)
print("ADMISSIONS BY CHRONIC CONDITION")
print("=" * 70)

display(chronic_admissions)

print("\nTotal Admissions:", chronic_admissions['admission_count'].sum())
```

```python
# Average length of stay by diagnosis

los_by_diagnosis = (
    analytics_df.groupby('diagnosis')['length_of_stay']
    .agg(['count', 'mean', 'min', 'max'])
    .round(2)
    .reset_index()
    .sort_values('mean', ascending=False)
)

print("=" * 70)
print("LENGTH OF STAY BY DIAGNOSIS")
print("=" * 70)

display(los_by_diagnosis)
```

```python
# Analyze whether previous admissions are associated with length of stay

previous_admission_los = (
    analytics_df.groupby('previous_admissions')['length_of_stay']
    .agg(['count', 'mean', 'min', 'max'])
    .round(2)
    .reset_index()
)

print("=" * 70)
print("PREVIOUS ADMISSIONS VS LENGTH OF STAY")
print("=" * 70)

display(previous_admission_los)
```

```python
# Vital sign patterns by diagnosis

vitals_by_diagnosis = (
    analytics_df.groupby('diagnosis')
    .agg(
        avg_heart_rate=('avg_heart_rate', 'mean'),
        avg_oxygen_level=('avg_oxygen_level', 'mean'),
        avg_temperature=('avg_temperature', 'mean'),
        avg_systolic_bp=('avg_systolic_bp', 'mean'),
        avg_diastolic_bp=('avg_diastolic_bp', 'mean')
    )
    .round(2)
    .reset_index()
    .sort_values('avg_heart_rate', ascending=False)
)

print("=" * 70)
print("VITAL SIGN SUMMARY BY DIAGNOSIS")
print("=" * 70)

display(vitals_by_diagnosis)
```

```python
# Treatment patterns by diagnosis

treatment_by_diagnosis = (
    analytics_df.groupby('diagnosis')
    .agg(
        admission_count=('admission_id', 'count'),
        avg_treatment_count=('treatment_count', 'mean'),
        avg_procedure_count=('procedure_count', 'mean'),
        avg_medication_count=('medication_count', 'mean')
    )
    .round(2)
    .reset_index()
    .sort_values('avg_treatment_count', ascending=False)
)

print("=" * 70)
print("TREATMENT ANALYSIS BY DIAGNOSIS")
print("=" * 70)

display(treatment_by_diagnosis)
```

```python
# Identify admissions with potentially concerning average vital signs

analytics_df['vital_risk_flags'] = (
    (analytics_df['avg_heart_rate'] > 100) |
    (analytics_df['avg_oxygen_level'] < 92) |
    (analytics_df['avg_temperature'] > 100) |
    (analytics_df['avg_systolic_bp'] > 140) |
    (analytics_df['avg_diastolic_bp'] > 90)
).astype(int)

vital_risk_summary = (
    analytics_df['vital_risk_flags']
    .value_counts()
    .rename(index={0: 'No Flag', 1: 'Potential Risk Flag'})
    .reset_index()
)

vital_risk_summary.columns = ['vital_status', 'admission_count']

print("=" * 70)
print("POTENTIAL VITAL SIGN RISK FLAGS")
print("=" * 70)

display(vital_risk_summary)

print("\nTotal Admissions:", vital_risk_summary['admission_count'].sum())
```

```python
# Potential vital-sign risk flags by diagnosis

risk_by_diagnosis = (
    analytics_df.groupby('diagnosis')
    .agg(
        admission_count=('admission_id', 'count'),
        risk_flag_count=('vital_risk_flags', 'sum')
    )
    .reset_index()
)

risk_by_diagnosis['risk_flag_percentage'] = (
    risk_by_diagnosis['risk_flag_count']
    / risk_by_diagnosis['admission_count'] * 100
).round(2)

risk_by_diagnosis = risk_by_diagnosis.sort_values(
    'risk_flag_percentage',
    ascending=False
)

print("=" * 70)
print("POTENTIAL VITAL RISK FLAGS BY DIAGNOSIS")
print("=" * 70)

display(risk_by_diagnosis)
```

```python
# Create age groups and analyze admissions

analytics_df['age_group'] = pd.cut(
    analytics_df['age'],
    bins=[0, 30, 45, 60, 75, 120],
    labels=['18-30', '31-45', '46-60', '61-75', '76+'],
    right=True
)

age_group_analysis = (
    analytics_df.groupby('age_group', observed=False)
    .agg(
        admission_count=('admission_id', 'count'),
        avg_length_of_stay=('length_of_stay', 'mean'),
        risk_flag_count=('vital_risk_flags', 'sum')
    )
    .round(2)
    .reset_index()
)

age_group_analysis['risk_flag_percentage'] = (
    age_group_analysis['risk_flag_count']
    / age_group_analysis['admission_count'] * 100
).round(2)

print("=" * 70)
print("AGE GROUP ANALYSIS")
print("=" * 70)

display(age_group_analysis)
```

```python
# Compare potential vital risk flags for patients with and without chronic conditions

chronic_risk_analysis = (
    analytics_df.groupby('has_chronic_condition')
    .agg(
        admission_count=('admission_id', 'count'),
        risk_flag_count=('vital_risk_flags', 'sum'),
        avg_length_of_stay=('length_of_stay', 'mean')
    )
    .round(2)
    .reset_index()
)

chronic_risk_analysis['risk_flag_percentage'] = (
    chronic_risk_analysis['risk_flag_count']
    / chronic_risk_analysis['admission_count'] * 100
).round(2)

chronic_risk_analysis['chronic_status'] = chronic_risk_analysis[
    'has_chronic_condition'
].map({
    0: 'No Recorded Condition',
    1: 'Has Chronic Condition'
})

chronic_risk_analysis = chronic_risk_analysis[
    ['chronic_status', 'admission_count', 'risk_flag_count',
     'risk_flag_percentage', 'avg_length_of_stay']
]

print("=" * 70)
print("CHRONIC CONDITION VS VITAL RISK")
print("=" * 70)

display(chronic_risk_analysis)
```

```python
# Analyze admissions handled by each doctor

doctor_workload = (
    analytics_df.groupby('doctor_id')
    .agg(
        admission_count=('admission_id', 'count'),
        avg_length_of_stay=('length_of_stay', 'mean'),
        risk_flag_count=('vital_risk_flags', 'sum')
    )
    .round(2)
    .reset_index()
    .sort_values('admission_count', ascending=False)
)

print("=" * 70)
print("DOCTOR WORKLOAD ANALYSIS")
print("=" * 70)

display(doctor_workload)
```

```python
# Create a combined admission risk profile
# This is an analytical screening score, not a clinical diagnosis.

analytics_df['risk_factor_count'] = (
    analytics_df['vital_risk_flags']
    + analytics_df['has_chronic_condition']
    + (analytics_df['previous_admissions'] > 0).astype(int)
)

risk_profile = (
    analytics_df.groupby('risk_factor_count')
    .agg(
        admission_count=('admission_id', 'count'),
        avg_length_of_stay=('length_of_stay', 'mean')
    )
    .round(2)
    .reset_index()
)

print("=" * 70)
print("ADMISSION RISK PROFILE")
print("=" * 70)

display(risk_profile)

print("\nTotal Admissions:", risk_profile['admission_count'].sum())
```

```python
# Prepare admission-level dataset for predictive analysis

model_df = analytics_df[
    [
        'admission_id',
        'age',
        'gender',
        'has_chronic_condition',
        'previous_admissions',
        'length_of_stay',
        'avg_heart_rate',
        'avg_oxygen_level',
        'avg_temperature',
        'avg_systolic_bp',
        'avg_diastolic_bp',
        'treatment_count',
        'procedure_count',
        'medication_count',
        'vital_risk_flags'
    ]
].copy()

print("=" * 70)
print("PREDICTIVE MODELING DATASET")
print("=" * 70)

print("Shape:", model_df.shape)
print("\nMissing values:")
display(model_df.isnull().sum())

print("\nPreview:")
display(model_df.head())
```

```python
# Check whether an actual readmission target is available

possible_target_columns = [
    col for col in analytics_df.columns
    if any(keyword in col.lower() for keyword in ['readmission', 'readmit', 'risk_level', 'risk_score'])
]

print("=" * 70)
print("READMISSION TARGET AVAILABILITY CHECK")
print("=" * 70)

if possible_target_columns:
    print("Potential target columns found:")
    print(possible_target_columns)
else:
    print("No actual readmission/risk target column is available in the supplied datasets.")

print("\nAvailable analytics columns:")
print(analytics_df.columns.tolist())
```

```python
# Create a project-defined analytical risk score
# This is NOT a clinical diagnosis or an observed readmission outcome.

risk_score_df = analytics_df[
    [
        'admission_id',
        'age',
        'previous_admissions',
        'has_chronic_condition',
        'vital_risk_flags',
        'length_of_stay',
        'avg_oxygen_level',
        'avg_heart_rate'
    ]
].copy()

# Assign points to selected risk indicators
risk_score_df['age_points'] = pd.cut(
    risk_score_df['age'],
    bins=[0, 45, 60, 75, 120],
    labels=[0, 1, 2, 3],
    include_lowest=True
).astype(int)

risk_score_df['previous_admission_points'] = (
    risk_score_df['previous_admissions'].clip(upper=2)
)

risk_score_df['chronic_condition_points'] = (
    risk_score_df['has_chronic_condition']
)

risk_score_df['vital_risk_points'] = (
    risk_score_df['vital_risk_flags']
)

risk_score_df['risk_score'] = (
    risk_score_df['age_points']
    + risk_score_df['previous_admission_points']
    + risk_score_df['chronic_condition_points']
    + risk_score_df['vital_risk_points']
)

print("=" * 70)
print("ANALYTICAL RISK SCORE")
print("=" * 70)

print("Score range:",
      risk_score_df['risk_score'].min(),
      "to",
      risk_score_df['risk_score'].max())

display(
    risk_score_df[
        ['admission_id', 'age', 'previous_admissions',
         'has_chronic_condition', 'vital_risk_flags',
         'risk_score']
    ].head(10)
)
```

```python
# Categorize the analytical risk score into Low, Medium, and High levels

risk_score_df['risk_level'] = pd.cut(
    risk_score_df['risk_score'],
    bins=[-1, 2, 4, 7],
    labels=['Low', 'Medium', 'High']
)

print("=" * 70)
print("ANALYTICAL RISK LEVEL DISTRIBUTION")
print("=" * 70)

print(risk_score_df['risk_level'].value_counts().sort_index())

print("\nRisk level percentages:")
print(
    (risk_score_df['risk_level'].value_counts(normalize=True).sort_index() * 100)
    .round(2)
)
```

```python
# Add analytical risk score and risk level to the main analytics dataset

analytics_df = analytics_df.merge(
    risk_score_df[
        ['admission_id', 'risk_score', 'risk_level']
    ],
    on='admission_id',
    how='left'
)

print("=" * 70)
print("FINAL ANALYTICS DATASET")
print("=" * 70)

print("\nShape:", analytics_df.shape)

print("\nMissing values in risk fields:")
print(
    analytics_df[
        ['risk_score', 'risk_level']
    ].isnull().sum()
)

print("\nRisk Level Distribution:")
print(
    analytics_df['risk_level']
    .value_counts()
    .sort_index()
)

print("\nFinal columns:")
print(analytics_df.columns.tolist())
```

```python
# Final data quality validation before MySQL export

print("=" * 70)
print("FINAL DATA QUALITY CHECK")
print("=" * 70)

print("\n1. Dataset shape:")
print(analytics_df.shape)

print("\n2. Duplicate admission IDs:")
print(analytics_df['admission_id'].duplicated().sum())

print("\n3. Missing values:")
print("Total missing values:", analytics_df.isnull().sum().sum())

print("\n4. Risk score range:")
print(
    analytics_df['risk_score'].min(),
    "to",
    analytics_df['risk_score'].max()
)

print("\n5. Risk level counts:")
print(analytics_df['risk_level'].value_counts().sort_index())

print("\n6. Admission date range:")
print(
    analytics_df['admission_date'].min(),
    "to",
    analytics_df['admission_date'].max()
)

print("\n7. Patient IDs without matching patient records:")
print(
    (~analytics_df['patient_id'].isin(patients['patient_id'])).sum()
)

print("\n8. Doctor IDs without matching doctor records:")
print(
    (~analytics_df['doctor_id'].isin(doctors['doctor_id'])).sum()
)

print("\nFinal validation completed.")
```
